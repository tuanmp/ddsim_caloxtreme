```python
import pandas as pd
import numpy as np
```

```python
min_e = 10
max_e = 500
step_e = 10

def generate_e_values(min_e, max_e, step_e):
    return np.arange(min_e, max_e + step_e, step_e)
```

```python
num_nodes = 10
num_processes = 100

events_per_energy = 100000
time_per_event = 0.007
storage_per_event = 0.138
train_energies = generate_e_values(20, max_e, 20)

val_energies = generate_e_values(30, max_e, 20)

train_df = pd.DataFrame({
    'incident_energy':  train_energies,
    'events': [events_per_energy] * len(train_energies),
    "time" : [events_per_energy * time_per_event] * len(train_energies),
    "storage": [events_per_energy * storage_per_event / 1000] * len(train_energies) 
})

train_df
```
